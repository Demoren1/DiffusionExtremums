```python
import numpy as np
import torch
from matplotlib import pyplot as plt
```

### Dataset Generation

```python
k = 0.7, 1.5
b = 1, -1

x = np.linspace(-40, 40, 1000)
y = np.sin(k[0] * x + b[0]) + np.sin(k[1] * x + b[1])

plt.figure(figsize=(12, 4), dpi=120)
plt.plot(x, y, color='#2563eb', linewidth=1.5)
plt.fill_between(x, y, alpha=0.1, color='#2563eb')
plt.axhline(0, color='gray', linewidth=0.5, alpha=0.5)
plt.grid(alpha=0.3, linestyle='--')
plt.title(f'sin({k[0]}x+{b[0]}) + sin({k[1]}x-{b[1]})', fontweight='bold')
plt.xlabel('x'); plt.ylabel('y')
plt.tight_layout()
plt.show()
```
